## 2. Data Types
Data science me data ke alag-alag types: Numerical (discrete, continuous), Categorical (nominal, ordinal), Boolean, Date-Time aur Text.

```python
import pandas as pd

# Ek sample dataset jisme alag-alag data types hain
data = {
    "Employee_ID": [101, 102, 103, 104, 105],                                  # Numerical - Discrete (ID/count)
    "Salary": [30500.50, 45000.75, 28000.00, 52000.25, 39000.00],              # Numerical - Continuous
    "Department": ["IT", "HR", "IT", "Finance", "HR"],                         # Categorical - Nominal
    "Performance": ["Good", "Excellent", "Average", "Good", "Excellent"],      # Categorical - Ordinal
    "Is_Permanent": [True, True, False, True, False],                          # Boolean
    "Joining_Date": ["2024-01-15", "2023-06-10", "2025-03-01", "2022-11-20", "2025-08-05"],  # Date-Time
    "Feedback": ["Hard working", "Great team player", "Needs improvement", "Reliable", "Creative thinker"]  # Text
}

df = pd.DataFrame(data)

# Date column ko proper datetime type me convert karo
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

# Ordinal column ko order ke saath category type me convert karo
df["Performance"] = pd.Categorical(
    df["Performance"],
    categories=["Average", "Good", "Excellent"],
    ordered=True
)

print(df)
print("\n--- Data Types of each column ---")
print(df.dtypes)
```

```python
# Har column ki classification ek table me
classification = pd.DataFrame({
    "Column": ["Employee_ID", "Salary", "Department", "Performance",
               "Is_Permanent", "Joining_Date", "Feedback"],
    "Pandas dtype": ["int64", "float64", "object", "category",
                     "bool", "datetime64", "object"],
    "Data Type": ["Numerical", "Numerical", "Categorical", "Categorical",
                  "Boolean", "Date-Time", "Text"],
    "Sub-type": ["Discrete", "Continuous", "Nominal", "Ordinal",
                 "Binary", "Time-series", "Unstructured"]
})
print(classification)

# Proof: ordinal data me order hota hai, nominal me nahi
print("\n--- Ordinal Check ---")
print("Excellent > Good ?", df["Performance"].iloc[1] > df["Performance"].iloc[0])

# Nominal data me sirf counting hoti hai
print("\n--- Nominal Count ---")
print(df["Department"].value_counts())
```

## 3. Statistical Foundations
Descriptive statistics: central tendency (mean, median, mode), spread (variance, standard deviation, range, quartiles).

```python
import numpy as np
import pandas as pd

# Sample data: 10 employees ki monthly salary
salaries = [28000, 30000, 30000, 32000, 35000, 38000, 40000, 45000, 52000, 120000]
s = pd.Series(salaries)

print("--- Central Tendency ---")
print("Mean   :", s.mean())
print("Median :", s.median())
print("Mode   :", s.mode()[0])

print("\n--- Spread ---")
print("Range             :", s.max() - s.min())
print("Variance          :", round(s.var(), 2))
print("Standard Deviation:", round(s.std(), 2))

print("\n--- Quartiles ---")
q1 = s.quantile(0.25)
q2 = s.quantile(0.50)
q3 = s.quantile(0.75)
print("Q1 (25%):", q1)
print("Q2 (50%):", q2)
print("Q3 (75%):", q3)
print("IQR      :", q3 - q1)

print("\n--- Quick Summary ---")
print(s.describe())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------- Correlation ----------
print("--- Correlation ---")
study = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks":       [35, 40, 50, 55, 65, 70, 80, 85]
})
corr = study["Study_Hours"].corr(study["Marks"])
print("Correlation (Study Hours vs Marks):", round(corr, 3))

# ---------- Normal Distribution ----------
print("\n--- Normal Distribution ---")
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)
print("Mean:", round(data.mean(), 2))
print("Std Dev:", round(data.std(), 2))

# Empirical rule (68-95-99.7)
m, sd = data.mean(), data.std()
for k in [1, 2, 3]:
    pct = ((data > m - k*sd) & (data < m + k*sd)).mean() * 100
    print(f"Within {k} std dev: {round(pct, 1)}%")

# ---------- Plots ----------
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].scatter(study["Study_Hours"], study["Marks"], color="orange")
ax[0].set_title("Correlation: Study Hours vs Marks")
ax[0].set_xlabel("Study Hours")
ax[0].set_ylabel("Marks")

ax[1].hist(data, bins=30, color="steelblue", edgecolor="white")
ax[1].axvline(m, color="red", linestyle="--", label="Mean")
ax[1].set_title("Normal Distribution")
ax[1].legend()

plt.tight_layout()
plt.show()
```

```python
from scipy import stats
import numpy as np

# Scenario: 2 groups ke marks. Kya training se marks badhe?
# Group A: bina training, Group B: training ke baad
group_a = [72, 75, 78, 80, 68, 74, 77, 79, 73, 76]
group_b = [82, 85, 79, 88, 84, 86, 81, 90, 83, 87]

print("Group A Mean:", np.mean(group_a))
print("Group B Mean:", np.mean(group_b))

# Hypothesis
# H0 (Null): Dono groups ka mean same hai (training ka koi asar nahi)
# H1 (Alternative): Dono groups ka mean alag hai (training ka asar hai)

t_stat, p_value = stats.ttest_ind(group_a, group_b)

print("\nT-statistic:", round(t_stat, 3))
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("\nResult: p-value < 0.05, isliye H0 reject. Training ka significant asar hai.")
else:
    print("\nResult: p-value >= 0.05, isliye H0 reject nahi hota. Koi significant fark nahi.")
```

### 4.1 Data Cleaning

```python
import pandas as pd
import numpy as np

# Ek "gandha" (messy) dataset
raw = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan", "Neha", "Riya", "Vikas", "Pooja"],
    "Age": [25, 28, np.nan, 30, 28, 120, 26],
    "City": ["Ranchi", "ranchi ", "Patna", "Delhi", "ranchi ", "Patna", None],
    "Salary": [30000, 45000, 28000, np.nan, 45000, 50000, 41000]
})
print("--- Raw Data ---")
print(raw)
print("\nMissing values:\n", raw.isnull().sum())
print("Duplicate rows:", raw.duplicated().sum())

# ---------- Cleaning ----------
df = raw.copy()
df["City"] = df["City"].str.strip().str.title()          # spaces hatao, format theek karo
df = df.drop_duplicates()                                # duplicate rows hatao
df["Age"] = df["Age"].fillna(df["Age"].median())         # missing Age -> median
df["Salary"] = df["Salary"].fillna(df["Salary"].median())  # missing Salary -> median
df["City"] = df["City"].fillna("Unknown")                # missing City -> Unknown
df = df[df["Age"] <= 100]                                # unrealistic Age (outlier) hatao
df = df.reset_index(drop=True)

print("\n--- Cleaned Data ---")
print(df)
print("\nMissing values after cleaning:", df.isnull().sum().sum())
print("Rows before:", len(raw), "| Rows after:", len(df))
```

4.2 Exploratory Data Analysis (EDA) aur 4.3 Reproducibility

### 4.4 Documentation aur Data **Ethics**

```python
# ---------- 4.4 Documentation: function ke saath docstring ----------
def salary_summary(data, column):
    """
    Kisi bhi numeric column ka basic summary return karta hai.

    Parameters:
        data (DataFrame): jis dataset pe kaam karna hai
        column (str): numeric column ka naam

    Returns:
        dict: mean, median, min aur max
    """
    return {
        "mean": float(data[column].mean()),
        "median": float(data[column].median()),
        "min": float(data[column].min()),
        "max": float(data[column].max())
    }

print("--- Docstring ---")
print(salary_summary.__doc__)

print("--- Function Output ---")
print(salary_summary(df, "Salary"))

# ---------- Data Ethics: personal info hatana (anonymization) ----------
print("\n--- Anonymized Data ---")
df_anon = df.drop(columns=["Name"])
print(df_anon)
```
